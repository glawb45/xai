Before we start with our attacks, we should verify the performance of our model. As ImageNet has 1000 classes, simply looking at the accuracy is not sufficient to tell the performance of a model. Imagine a model that always predicts the true label as the second-highest class in its softmax output. Although we would say it recognizes the object in the image, it achieves an accuracy of 0. In ImageNet with 1000 classes, there is not always one clear label we can assign an image to. This is why for image classifications over so many classes, a common alternative metric is "Top-5 accuracy", which tells us how many times the true label has been within the 5 most-likely predictions of the model. As models usually perform quite well on those, we report the error (1 - accuracy) instead of the accuracy:

```python
def eval_model(dataset_loader, img_func=None):
    tp, tp_5, counter = 0.0, 0.0, 0.0
    for imgs, labels in tqdm(dataset_loader, desc="Validating..."):
        imgs = imgs.to(device)
        labels = labels.to(device)
        if img_func is not None:
            imgs = img_func(imgs, labels)
        with torch.no_grad():
            preds = pretrained_model(imgs)
        tp += (preds.argmax(dim=-1) == labels).sum()
        tp_5 += (preds.topk(5, dim=-1)[1] == labels[..., None]).any(dim=-1).sum()
        counter += preds.shape[0]
    acc = tp.float().item() / counter
    top5 = tp_5.float().item() / counter
    print(f"Top-1 error: {(100.0 * (1 - acc)):4.2f}%")
    print(f"Top-5 error: {(100.0 * (1 - top5)):4.2f}%")
    return acc, top5
```

The function below plots an image along with a bar diagram of its predictions. We also prepare it to show adversarial examples for later applications.

```python
def show_prediction(img, label, pred, K=5, adv_img=None, noise=None):

    if isinstance(img, torch.Tensor):
        # Tensor image to numpy
        img = img.cpu().permute(1, 2, 0).numpy()
        img = (img * NORM_STD[None, None]) + NORM_MEAN[None, None]
        img = np.clip(img, a_min=0.0, a_max=1.0)
        label = label.item()

    # Plot on the left the image with the true label as title.
    # On the right, have a horizontal bar plot with the top k predictions including probabilities
    if noise is None or adv_img is None:
        fig, ax = plt.subplots(
            1, 2, figsize=(10, 2), gridspec_kw={"width_ratios": [1, 1]}
        )
    else:
        fig, ax = plt.subplots(
            1, 5, figsize=(12, 2), gridspec_kw={"width_ratios": [1, 1, 1, 1, 2]}
        )

    ax[0].imshow(img)
    ax[0].set_title(label_names[label])
    ax[0].axis("off")

    if adv_img is not None and noise is not None:
        # Visualize adversarial images
        adv_img = adv_img.cpu().permute(1, 2, 0).numpy()
        adv_img = (adv_img * NORM_STD[None, None]) + NORM_MEAN[None, None]
        adv_img = np.clip(adv_img, a_min=0.0, a_max=1.0)
        ax[1].imshow(adv_img)
        ax[1].set_title("Adversarial")
        ax[1].axis("off")
        # Visualize noise
        noise = noise.cpu().permute(1, 2, 0).numpy()
        noise = noise * 0.5 + 0.5  # Scale between 0 to 1
        ax[2].imshow(noise)
        ax[2].set_title("Noise")
        ax[2].axis("off")
        # buffer
        ax[3].axis("off")

    if abs(pred.sum().item() - 1.0) > 1e-4:
        pred = torch.softmax(pred, dim=-1)
    topk_vals, topk_idx = pred.topk(K, dim=-1)
    topk_vals, topk_idx = topk_vals.cpu().numpy(), topk_idx.cpu().numpy()
    ax[-1].barh(
        np.arange(K),
        topk_vals * 100.0,
        align="center",
        color=["C0" if topk_idx[i] != label else "C2" for i in range(K)],
    )
    ax[-1].set_yticks(np.arange(K))
    ax[-1].set_yticklabels([label_names[c] for c in topk_idx])
    ax[-1].invert_yaxis()
    ax[-1].set_xlabel("Confidence")
    ax[-1].set_title("Predictions")

    plt.show()
    plt.close()
```

### Adversarial Patches

Instead of changing every pixel by a little bit, we can also try to change a small part of the image into whatever values we would like. In other words, we will create a small image patch that covers a minor part of the original image but causes the model to confidentially predict a specific class we choose. This form of attack is an even bigger threat in real-world applications than FSGM. Imagine a network in an autonomous car that receives a live image from a camera. Another driver could print out a specific pattern and put it on the back of his/her vehicle to make the autonomous car believe that the car is actually a pedestrian. Meanwhile, humans would not notice it. [Tom Brown et al.](https://arxiv.org/pdf/1712.09665.pdf) proposed a way of learning such adversarial image patches robustly in 2017 and provided a short demonstration on [YouTube](https://youtu.be/i1sp4X57TL4). Interestingly, if you add a small picture of the target class (here *toaster*) to the original image, the model does not pick it up at all. A specifically designed patch, however, which only roughly looks like a toaster, can change the network's prediction instantaneously.

[![Adversarial patch in real world](https://img.youtube.com/vi/i1sp4X57TL4/0.jpg)](https://youtu.be/i1sp4X57TL4)

Let's take a closer look at how we can actually train such patches. We calculate gradients for the input, and update our adversarial input correspondingly. We do not calculate a gradient for every pixel. Instead, we replace parts of the input image with our patch and then calculate the gradients just for our patch. Secondly, we don't just do it for one image, but we want the patch to work with any possible image. Hence, we have a whole training loop where we train the patch using SGD. Lastly, image patches are usually designed to make the model predict a specific class, not just any other arbitrary class except the true label. For instance, we can try to create a patch for the class "toaster" and train the patch so that our pretrained model predicts the class "toaster" for any image with the patch in it.

[Brown et al.](https://arxiv.org/pdf/1712.09665.pdf) randomly rotated and scaled the patch during training before placing it at a random position in an input image. This makes the patch more robust to small changes and is necessary if we want to fool a neural network in a real-world application.

This tutorial only shows how to make the patch robust to the location in the image.

Here is where you could:
* Randomly rotate patch
* Randomly scale the patch
* Improve random location of patch in the image

```python
def place_patch(img, patch):
    for i in range(img.shape[0]):
        h_offset = np.random.randint(0, img.shape[2] - patch.shape[1] - 1)
        w_offset = np.random.randint(0, img.shape[3] - patch.shape[2] - 1)
        img[
            i,
            :,
            h_offset : h_offset + patch.shape[1],
            w_offset : w_offset + patch.shape[2],
        ] = patch_forward(patch)
    return img
```

The patch itself will be an `nn.Parameter` whose values are in the range between $-\infty$ and $\infty$. Images are, however, naturally limited in their range, and thus we write a small function that maps the parameter into the image value range of ImageNet:

```python
TENSOR_MEANS, TENSOR_STD = (
    torch.FloatTensor(NORM_MEAN)[:, None, None],
    torch.FloatTensor(NORM_STD)[:, None, None],
)


def patch_forward(patch):
    # Map patch values from [-infty,infty] to ImageNet min and max
    patch = (torch.tanh(patch) + 1 - 2 * TENSOR_MEANS) / (2 * TENSOR_STD)
    return patch
```

Before looking at the actual training code, we can write a small evaluation function. We evaluate the success of a patch by how many times we were able to fool the network into predicting our target class. A simple function for this is implemented below.

```python
def eval_patch(model, patch, val_loader, target_class):
    model.eval()
    tp, tp_5, counter = 0.0, 0.0, 0.0
    with torch.no_grad():
        for img, img_labels in tqdm(val_loader, desc="Validating...", leave=False):
            # For stability, place the patch at 4 random locations per image, and average the performance
            for _ in range(4):
                patch_img = place_patch(img, patch)
                patch_img = patch_img.to(device)
                img_labels = img_labels.to(device)
                pred = model(patch_img)
                # In the accuracy calculation, we need to exclude the images that are of our target class
                # as we would not "fool" the model into predicting those
                tp += torch.logical_and(
                    pred.argmax(dim=-1) == target_class, img_labels != target_class
                ).sum()
                tp_5 += torch.logical_and(
                    (pred.topk(5, dim=-1)[1] == target_class).any(dim=-1),
                    img_labels != target_class,
                ).sum()
                counter += (img_labels != target_class).sum()
    acc = tp / counter
    top5 = tp_5 / counter
    return acc, top5
```

Finally, we can look at the training loop. Given a model to fool, a target class to design the patch for, and a size $k$ of the patch in the number of pixels, we first start by creating a parameter of size $3\times k\times k$. These are the only parameters we will train, and the network itself remains untouched. We use a simple SGD optimizer with momentum to minimize the classification loss of the model given the patch in the image. While we first start with a very high loss due to the good initial performance of the network, the loss quickly decreases once we start changing the patch. In the end, the patch will represent patterns that are characteristic of the class. For instance, if we would want the model to predict a "goldfish" in every image, we would expect the pattern to look somewhat like a goldfish. Over the iterations, the model finetunes the pattern and, hopefully, achieves a high fooling accuracy.

```python
def patch_attack(model, target_class, patch_size=64, num_epochs=5):
    # Leave a small set of images out to check generalization
    # In most of our experiments, the performance on the hold-out data points
    # was as good as on the training set. Overfitting was little possible due
    # to the small size of the patches.
    train_set, val_set = torch.utils.data.random_split(dataset, [4500, 500])
    train_loader = data.DataLoader(
        train_set, batch_size=32, shuffle=True, drop_last=True, num_workers=8
    )
    val_loader = data.DataLoader(
        val_set, batch_size=32, shuffle=False, drop_last=False, num_workers=4
    )

    # Create parameter and optimizer
    if not isinstance(patch_size, tuple):
        patch_size = (patch_size, patch_size)
    patch = nn.Parameter(
        torch.zeros(3, patch_size[0], patch_size[1]), requires_grad=True
    )
    optimizer = torch.optim.SGD([patch], lr=1e-1, momentum=0.8)
    loss_module = nn.CrossEntropyLoss()

    # Training loop
    for epoch in range(num_epochs):
        t = tqdm(train_loader, leave=False)
        for img, _ in t:
            img = place_patch(img, patch)
            img = img.to(device)
            pred = model(img)
            labels = torch.zeros(
                img.shape[0], device=pred.device, dtype=torch.long
            ).fill_(target_class)
            loss = loss_module(pred, labels)
            optimizer.zero_grad()
            loss.mean().backward()
            optimizer.step()
            t.set_description(f"Epoch {epoch}, Loss: {loss.item():4.2f}")

    # Final validation
    acc, top5 = eval_patch(model, patch, val_loader, target_class)

    return patch.data, {"acc": acc.item(), "top5": top5.item()}
```

To get some experience with what to expect from an adversarial patch attack, we want to train multiple patches for different classes. As the training of a patch can take one or two minutes on a GPU, we have provided a couple of pre-trained patches including their results on the full dataset. The results are saved in a JSON file, which is loaded below.

```python
# Load evaluation results of the pretrained patches
json_results_file = os.path.join(CHECKPOINT_PATH, "patch_results.json")
json_results = {}
if os.path.isfile(json_results_file):
    with open(json_results_file, "r") as f:
        json_results = json.load(f)


# If you train new patches, you can save the results via calling this function
def save_results(patch_dict):
    result_dict = {
        cname: {
            psize: [
                t.item() if isinstance(t, torch.Tensor) else t
                for t in patch_dict[cname][psize]["results"]
            ]
            for psize in patch_dict[cname]
        }
        for cname in patch_dict
    }
    with open(os.path.join(CHECKPOINT_PATH, "patch_results.json"), "w") as f:
        json.dump(result_dict, f, indent=4)
```

Additionally, we implement a function to train and evaluate patches for a list of classes and patch sizes. The pretrained patches include the classes *toaster*, *goldfish*, *school bus*, *lipstick*, and *pineapple*. We chose the classes arbitrarily to cover multiple domains (animals, vehicles, fruits, devices, etc.). We trained each class for three different patch sizes: $32\times32$ pixels, $48\times48$ pixels, and $64\times64$ pixels. We can load them in the two cells below.

```python
def get_patches(class_names, patch_sizes):
    result_dict = dict()

    # Loop over all classes and patch sizes
    for name in class_names:
        result_dict[name] = dict()
        for patch_size in patch_sizes:
            c = label_names.index(name)
            file_name = os.path.join(CHECKPOINT_PATH, f"{name}_{patch_size}_patch.pt")
            # Load patch if pretrained file exists, otherwise start training
            if not os.path.isfile(file_name):
                patch, val_results = patch_attack(
                    pretrained_model,
                    target_class=c,
                    patch_size=patch_size,
                    num_epochs=5,
                )
                print(f"Validation results for {name} and {patch_size}:", val_results)
                torch.save(patch, file_name)
            else:
                patch = torch.load(file_name)
            # Load evaluation results if exist, otherwise manually evaluate the patch
            if name in json_results:
                results = json_results[name][str(patch_size)]
            else:
                results = eval_patch(
                    pretrained_model, patch, data_loader, target_class=c
                )

            # Store results and the patches in a dict for better access
            result_dict[name][patch_size] = {"results": results, "patch": patch}

    return result_dict
```

## Creative Extension: A Rotation-Keyed Patch Alphabet for Secret Messages

The usual way to handle rotations (e.g. Brown et al.) is to make a patch **invariant** to them: it should predict the same class no matter how it's turned. Here we do the opposite and make the rotation **carry information**. A single patch is trained so that the model predicts a *different* class for each of its four orientations:

| Rotation | 0° | 90° | 180° | 270° |
|---|---|---|---|---|
| Predicted class | class A | class B | class C | class D |

One printed sticker therefore has four "personalities" and encodes 2 bits just by how you turn it. This covers all three of the suggested directions:

* **Rotations:** each patch is trained with random location, scale and small rotation jitter (Expectation over Transformation), and the 90° rotations are used as keys.
* **Combining patches:** four targeted patches are packed into one set of pixels. We check how much the keys interfere, and what the model sees *between* two keys (e.g. at 45°).
* **Secret message:** we assign letters to (patch, rotation) slots and send a message as a series of patched images. The **decoder is the ResNet itself**: read off the first letter of the predicted ImageNet class. A human sees colorful noise on random photos, while anyone with a ResNet34 reads the text.

Finally, we treat the patches as a noisy communication channel and measure the character error rate under rotation/scale noise, with and without a simple repetition code.

### Placing a rotated and scaled patch

The original `place_patch` can only paste an axis-aligned patch. Below is a drop-in replacement that uses a single differentiable affine warp (`F.affine_grid` + `F.grid_sample`) to rotate, scale and translate the patch for the whole batch at once. A warped mask of ones handles the transparent corners of a rotated square, so the patch is alpha-composited onto the image. Everything stays on the GPU, so it is also faster than the per-image Python loop.

```python
TENSOR_MEANS_DEV, TENSOR_STD_DEV = TENSOR_MEANS.to(device), TENSOR_STD.to(device)


def patch_forward_dev(patch):
    # Same as patch_forward, but keeps everything on the GPU
    return (torch.tanh(patch) + 1 - 2 * TENSOR_MEANS_DEV) / (2 * TENSOR_STD_DEV)


def place_patch_affine(imgs, patch_img, angles, scales=None, centers=None):
    """
    Pastes a rotated/scaled copy of `patch_img` into every image of the batch.
    imgs:      (B,3,H,W) normalized images (H == W)
    patch_img: (3,p,p) normalized patch (output of patch_forward_dev)
    angles:    (B,) rotation in degrees (counter-clockwise)
    scales:    (B,) size relative to the patch's native size (default 1)
    centers:   (B,2) patch center in [-1,1] image coordinates (default: random, fully inside the image)
    """
    B, _, H, W = imgs.shape
    p = patch_img.shape[-1]
    if scales is None:
        scales = torch.ones(B, device=imgs.device)
    half = (
        scales * p / W
    )  # Half-width of the patch in normalized [-1,1] image coordinates
    if centers is None:
        # Keep the patch fully inside the image for any rotation (half-diagonal = sqrt(2) * half-width)
        lim = (1 - half * math.sqrt(2)).clamp(min=0)
        centers = (torch.rand(B, 2, device=imgs.device) * 2 - 1) * lim[:, None]
    rad = angles * math.pi / 180
    cos, sin = torch.cos(rad), torch.sin(rad)
    # affine_grid maps output (image) coordinates to input (patch) coordinates
    rot = torch.stack(
        [torch.stack([cos, -sin], dim=-1), torch.stack([sin, cos], dim=-1)], dim=-2
    )
    A = rot / half[:, None, None]
    t = -(A @ centers[:, :, None])
    theta = torch.cat([A, t], dim=2)
    grid = F.affine_grid(theta, (B, 3, H, W), align_corners=False)
    src = patch_img[None].expand(B, -1, -1, -1)
    warped = F.grid_sample(src, grid, align_corners=False, padding_mode="zeros")
    mask = F.grid_sample(
        torch.ones_like(src[:, :1]), grid, align_corners=False, padding_mode="zeros"
    )
    return imgs * (1 - mask) + warped * mask


def tensor_to_display(img):
    # Normalized (3,H,W) tensor -> (H,W,3) numpy image in [0,1]
    img = img.detach().cpu().permute(1, 2, 0).numpy()
    return np.clip(img * NORM_STD[None, None] + NORM_MEAN[None, None], 0.0, 1.0)
```

### The codebook: letters → (patch, rotation) → ImageNet class

The decoding rule is public and simple: **take the first letter of the predicted class name**. A space is sent with a class whose name starts with "space" (e.g. *space shuttle*). For every distinct symbol in the message we pick a visually distinctive ImageNet class that starts with that letter, and assign the symbols, four at a time, to the rotations of each patch.

Change `MESSAGE` to send something else. Every 4 distinct symbols cost one extra patch to train. Avoid letters that no ImageNet class starts with (e.g. X).

```python
from collections import Counter

MESSAGE = "HELLO DUKE"  # Change me! Letters A-Z and spaces

ROTATION_ANGLES = [0, 90, 180, 270]
NUM_ROTATIONS = len(ROTATION_ANGLES)


def decode_class(class_idx):
    # The public decoding rule: first letter of the predicted class ("space ..." classes encode a space)
    name = label_names[class_idx]
    return " " if name.lower().startswith("space") else name[0].upper()


# Visually distinctive objects tend to be easier patch targets than e.g. one of 120 dog breeds
PREFERRED_CLASSES = {
    "A": ["ambulance", "acorn", "airliner"],
    "B": ["banana", "balloon", "broccoli"],
    "C": ["cheeseburger", "castle", "canoe"],
    "D": ["daisy", "dumbbell"],
    "E": ["espresso", "envelope"],
    "F": ["flamingo", "fire truck", "fire engine"],
    "G": ["goldfish", "golf ball"],
    "H": ["hotdog", "hot dog", "hen"],
    "I": ["ice cream", "iPod"],
    "J": ["jellyfish", "jack-o'-lantern"],
    "K": ["koala", "king penguin"],
    "L": ["lemon", "lipstick", "ladybug"],
    "M": ["monarch butterfly", "mushroom"],
    "N": ["nail", "necklace"],
    "O": ["orange", "ostrich"],
    "P": ["pineapple", "pizza"],
    "Q": ["quill"],
    "R": ["red wine", "rugby ball"],
    "S": ["school bus", "strawberry"],
    "T": ["toaster", "tennis ball"],
    "U": ["umbrella", "unicycle"],
    "V": ["volcano", "violin"],
    "W": ["wine bottle", "waffle iron"],
    "Y": ["yawl"],
    "Z": ["zebra"],
    " ": ["space shuttle", "space heater"],
}


def pick_class(symbol):
    for name in PREFERRED_CLASSES.get(symbol, []):
        if name in label_names and decode_class(label_names.index(name)) == symbol:
            return label_names.index(name)
    # Fallback: the first class that the decoder maps to this symbol
    for idx in range(len(label_names)):
        if decode_class(idx) == symbol:
            return idx
    raise ValueError(
        f"No ImageNet class decodes to {symbol!r}, please choose a different message."
    )


symbols = list(
    dict.fromkeys(MESSAGE.upper())
)  # Distinct symbols in order of appearance
SYMBOL_TO_CLASS = {s: pick_class(s) for s in symbols}
# Slot j -> patch j // 4, rotation j % 4
SYMBOL_TO_SLOT = {
    s: (j // NUM_ROTATIONS, j % NUM_ROTATIONS) for j, s in enumerate(symbols)
}
SLOT_TO_SYMBOL = {slot: s for s, slot in SYMBOL_TO_SLOT.items()}

# patch_targets[p][k] = class that patch p should trigger when rotated by ROTATION_ANGLES[k] (None = unused)
patch_targets = [
    [None] * NUM_ROTATIONS for _ in range(math.ceil(len(symbols) / NUM_ROTATIONS))
]
for s, (p_idx, r_idx) in SYMBOL_TO_SLOT.items():
    patch_targets[p_idx][r_idx] = SYMBOL_TO_CLASS[s]

print(
    f"Message: {MESSAGE!r} -> {len(symbols)} distinct symbols -> {len(patch_targets)} patches\n"
)
for s, (p_idx, r_idx) in SYMBOL_TO_SLOT.items():
    print(
        f"  {s!r}: patch {p_idx} @ {ROTATION_ANGLES[r_idx]:3d}°  ->  {label_names[SYMBOL_TO_CLASS[s]]}"
    )
```

### Training a rotation-keyed patch

For one patch $\delta$ with targets $y_0, \dots, y_3$, we minimize

$$\mathcal{L}(\delta) = \sum_{k} \; \mathbb{E}_{x,\ \ell,\ s,\ \epsilon}\Big[\mathrm{CE}\big(f(A(x, \delta;\ \ell,\ s,\ 90^\circ k + \epsilon)),\ y_k\big)\Big]$$

where $A$ pastes the patch into image $x$ at a random location $\ell$, scale $s \in [0.9, 1.1]$ and angle $90^\circ k + \epsilon$ with jitter $\epsilon \in [-10^\circ, 10^\circ]$. The jitter makes the patch robust to imperfect placement, but it must stay well below $45^\circ$, otherwise neighboring keys overlap and contradict each other. In practice, all four rotated copies of a batch are stacked into one forward pass of $4 \times 32$ images.

Next to the usual fooling accuracy, we also measure **crosstalk**: how often a rotation triggers one of the patch's *other* keys instead of its own.

```python
def rotation_keyed_patch_attack(
    model,
    targets,
    train_loader,
    patch_size=64,
    num_epochs=6,
    angle_jitter=10.0,
    scale_range=(0.9, 1.1),
):
    # Rotating the patch by ROTATION_ANGLES[k] should make the model predict targets[k]
    active = [k for k, t in enumerate(targets) if t is not None]
    patch = nn.Parameter(
        torch.zeros(3, patch_size, patch_size, device=device), requires_grad=True
    )
    optimizer = torch.optim.SGD([patch], lr=1e-1, momentum=0.8)
    loss_module = nn.CrossEntropyLoss()

    for epoch in range(num_epochs):
        t = tqdm(train_loader, leave=False)
        for img, _ in t:
            img = img.to(device)
            B = img.shape[0]
            # One copy of the batch per active rotation, all in a single forward pass
            imgs = img.repeat(len(active), 1, 1, 1)
            base_angles = torch.tensor(
                [float(ROTATION_ANGLES[k]) for k in active], device=device
            ).repeat_interleave(B)
            labels = torch.tensor(
                [targets[k] for k in active], device=device
            ).repeat_interleave(B)
            angles = base_angles + (torch.rand_like(base_angles) * 2 - 1) * angle_jitter
            scales = torch.empty_like(base_angles).uniform_(*scale_range)
            imgs = place_patch_affine(imgs, patch_forward_dev(patch), angles, scales)
            loss = loss_module(model(imgs), labels)
            optimizer.zero_grad()
            loss.backward()
            optimizer.step()
            t.set_description(f"Epoch {epoch}, Loss: {loss.item():4.2f}")

    return patch.data


@torch.no_grad()
def eval_rotation_keyed_patch(
    model, patch, val_loader, targets, angle_jitter=0.0, scale_range=(1.0, 1.0)
):
    # Per rotation: top-1/top-5 fooling rate, and crosstalk (top-1 is one of the patch's OTHER keys)
    model.eval()
    patch_img = patch_forward_dev(patch.to(device))
    stats = {
        k: {"acc": 0.0, "top5": 0.0, "crosstalk": 0.0, "n": 0}
        for k, t in enumerate(targets)
        if t is not None
    }
    for img, img_labels in tqdm(val_loader, desc="Validating...", leave=False):
        img, img_labels = img.to(device), img_labels.to(device)
        B = img.shape[0]
        for k in stats:
            angles = (
                ROTATION_ANGLES[k]
                + (torch.rand(B, device=device) * 2 - 1) * angle_jitter
            )
            scales = torch.empty(B, device=device).uniform_(*scale_range)
            pred = model(place_patch_affine(img, patch_img, angles, scales))
            top1 = pred.argmax(dim=-1)
            # As before, exclude images that already belong to the target class
            valid = img_labels != targets[k]
            other_keys = torch.tensor(
                [c for j, c in enumerate(targets) if c is not None and j != k],
                device=device,
                dtype=torch.long,
            )
            stats[k]["acc"] += ((top1 == targets[k]) & valid).sum().item()
            stats[k]["top5"] += (
                ((pred.topk(5, dim=-1)[1] == targets[k]).any(dim=-1) & valid)
                .sum()
                .item()
            )
            stats[k]["crosstalk"] += (torch.isin(top1, other_keys) & valid).sum().item()
            stats[k]["n"] += valid.sum().item()
    return {
        k: {m: s[m] / s["n"] for m in ("acc", "top5", "crosstalk")}
        for k, s in stats.items()
    }
```

Now we train (or load) one patch per group of four symbols. With a 64×64 patch and 6 epochs, each patch takes a few minutes on a Colab T4 GPU.

```python
ROTKEY_PATCH_SIZE = 64

rotkey_train_set, rotkey_val_set = data.random_split(
    dataset, [4500, 500], generator=torch.Generator().manual_seed(42)
)
rotkey_train_loader = data.DataLoader(
    rotkey_train_set, batch_size=32, shuffle=True, drop_last=True, num_workers=4
)
rotkey_val_loader = data.DataLoader(
    rotkey_val_set, batch_size=32, shuffle=False, drop_last=False, num_workers=4
)

rotkey_patches, rotkey_results = [], []
for p_idx, targets in enumerate(patch_targets):
    tag = "-".join(str(c) if c is not None else "x" for c in targets)
    file_name = os.path.join(
        CHECKPOINT_PATH, f"rotkey_{tag}_{ROTKEY_PATCH_SIZE}_patch.pt"
    )
    if os.path.isfile(file_name):
        patch = torch.load(file_name, map_location=device)
    else:
        print(
            f"Training patch {p_idx}: "
            + ", ".join(
                label_names[c] if c is not None else "(unused)" for c in targets
            )
        )
        patch = rotation_keyed_patch_attack(
            pretrained_model, targets, rotkey_train_loader, patch_size=ROTKEY_PATCH_SIZE
        )
        torch.save(patch, file_name)
    rotkey_patches.append(patch)
    rotkey_results.append(
        eval_rotation_keyed_patch(pretrained_model, patch, rotkey_val_loader, targets)
    )

table = []
for p_idx, (targets, results) in enumerate(zip(patch_targets, rotkey_results)):
    for k, r in results.items():
        table.append(
            [
                p_idx,
                f"{ROTATION_ANGLES[k]}°",
                repr(SLOT_TO_SYMBOL[(p_idx, k)]),
                label_names[targets[k]],
                f"{100 * r['acc']:.1f}%",
                f"{100 * r['top5']:.1f}%",
                f"{100 * r['crosstalk']:.1f}%",
            ]
        )
display(
    HTML(
        tabulate.tabulate(
            table,
            tablefmt="html",
            headers=[
                "Patch",
                "Rotation",
                "Symbol",
                "Target class",
                "Top-1",
                "Top-5",
                "Crosstalk",
            ],
        )
    )
)
```

Let's look at the patches. Each row is one patch pasted onto the same image, rotated to each of its four keys.

```python
carrier = rotkey_val_set[0][0].to(device)
fig, ax = plt.subplots(
    len(rotkey_patches),
    NUM_ROTATIONS,
    figsize=(NUM_ROTATIONS * 2.8, len(rotkey_patches) * 3.1),
    squeeze=False,
)
with torch.no_grad():
    for p_idx, (patch, targets) in enumerate(zip(rotkey_patches, patch_targets)):
        imgs = carrier[None].repeat(NUM_ROTATIONS, 1, 1, 1)
        angles = torch.tensor(ROTATION_ANGLES, device=device, dtype=torch.float)
        adv = place_patch_affine(
            imgs,
            patch_forward_dev(patch),
            angles,
            centers=torch.zeros(NUM_ROTATIONS, 2, device=device),
        )
        probs = F.softmax(pretrained_model(adv), dim=-1)
        for k in range(NUM_ROTATIONS):
            c = probs[k].argmax().item()
            key = label_names[targets[k]] if targets[k] is not None else "(unused)"
            ax[p_idx][k].imshow(tensor_to_display(adv[k]))
            ax[p_idx][k].set_title(
                f"Patch {p_idx} @ {ROTATION_ANGLES[k]}°, key: {key}\n"
                f"pred: {label_names[c]} ({100 * probs[k, c]:.0f}%) → {decode_class(c)!r}",
                fontsize=8,
            )
            ax[p_idx][k].axis("off")
plt.tight_layout()
plt.show()
```

### What happens *between* the keys?

The patch was only trained near 0°, 90°, 180° and 270° (±10°). By sweeping the angle continuously from 0° to 360° on the same set of images (fixed positions, so only the angle changes), we can see how the model hands control from one key to the next. Is there a sharp switch at 45°, a region where both classes compete, or a gap where the patch loses control and the image's true class comes back?

```python
@torch.no_grad()
def angle_sweep(patch, num_imgs=128, step=5):
    patch_img = patch_forward_dev(patch.to(device))
    imgs = torch.stack([rotkey_val_set[i][0] for i in range(num_imgs)]).to(device)
    true_labels = torch.tensor(
        [rotkey_val_set[i][1] for i in range(num_imgs)], device=device
    )
    lim = 1 - ROTKEY_PATCH_SIZE / imgs.shape[-1] * math.sqrt(2)
    centers = (
        torch.rand(num_imgs, 2, device=device) * 2 - 1
    ) * lim  # Fixed across angles
    angles = np.arange(0, 360 + step, step)
    probs, true_probs = [], []
    for a in angles:
        pred = F.softmax(
            pretrained_model(
                place_patch_affine(
                    imgs,
                    patch_img,
                    torch.full((num_imgs,), float(a), device=device),
                    centers=centers,
                )
            ),
            dim=-1,
        )
        probs.append(pred.mean(dim=0).cpu())
        true_probs.append(pred.gather(1, true_labels[:, None]).mean().item())
    return angles, torch.stack(probs), np.array(true_probs)


fig, ax = plt.subplots(
    len(rotkey_patches), 1, figsize=(10, 3.2 * len(rotkey_patches)), squeeze=False
)
for p_idx, (patch, targets) in enumerate(zip(rotkey_patches, patch_targets)):
    angles, probs, true_probs = angle_sweep(patch)
    a = ax[p_idx][0]
    for k, c in enumerate(targets):
        if c is not None:
            a.plot(
                angles,
                probs[:, c].numpy(),
                label=f"{ROTATION_ANGLES[k]}°: {label_names[c]} ({SLOT_TO_SYMBOL[(p_idx, k)]!r})",
            )
    a.plot(angles, true_probs, "k--", label="true class of image")
    for b in range(45, 360, 90):
        a.axvline(b, color="gray", alpha=0.3, linestyle=":")
    a.set_xticks(range(0, 361, 45))
    a.set_xlabel("Patch rotation (degrees)")
    a.set_ylabel("Mean probability")
    a.set_title(f"Patch {p_idx}: angle sweep")
    a.legend(loc="center left", bbox_to_anchor=(1.01, 0.5), fontsize=8)
plt.tight_layout()
plt.show()
```

### Sending the secret message

Every character of `MESSAGE` becomes one "frame": a random image from the validation set with the matching patch pasted in at the right rotation. The receiver runs each frame through ResNet34 and applies the first-letter rule. To someone without the model, the frames are just random photos with colorful stickers on them.

```python
@torch.no_grad()
def transmit(message, repeats=1, angle_jitter=0.0, scale_range=(1.0, 1.0)):
    # Encode -> paste onto random carrier images -> decode with the model. Each symbol is sent `repeats` times
    msg = message.upper()
    slots = [SYMBOL_TO_SLOT[s] for s in msg for _ in range(repeats)]
    idx = torch.randperm(len(rotkey_val_set))[: len(slots)]
    imgs = torch.stack([rotkey_val_set[i][0] for i in idx]).to(device)
    patch_idx = torch.tensor([p for p, _ in slots], device=device)
    angles = torch.tensor([float(ROTATION_ANGLES[r]) for _, r in slots], device=device)
    angles = angles + (torch.rand_like(angles) * 2 - 1) * angle_jitter
    scales = torch.empty_like(angles).uniform_(*scale_range)
    frames = imgs.clone()
    for p_idx, patch in enumerate(rotkey_patches):
        sel = patch_idx == p_idx
        if sel.any():
            frames[sel] = place_patch_affine(
                imgs[sel], patch_forward_dev(patch), angles[sel], scales[sel]
            )
    preds = pretrained_model(frames).argmax(dim=-1).tolist()
    symbols_rx = [decode_class(c) for c in preds]
    # Majority vote over the repeated frames of each character (a simple repetition code)
    decoded = "".join(
        Counter(symbols_rx[i * repeats : (i + 1) * repeats]).most_common(1)[0][0]
        for i in range(len(msg))
    )
    return decoded, frames, preds


decoded, frames, preds = transmit(MESSAGE)

ncols = min(len(MESSAGE), 5)
nrows = math.ceil(len(MESSAGE) / ncols)
fig, ax = plt.subplots(nrows, ncols, figsize=(ncols * 2.4, nrows * 2.8), squeeze=False)
for i in range(nrows * ncols):
    a = ax[i // ncols][i % ncols]
    a.axis("off")
    if i < len(MESSAGE):
        a.imshow(tensor_to_display(frames[i]))
        a.set_title(
            f"{label_names[preds[i]]}\n→ {decode_class(preds[i])!r}", fontsize=9
        )
plt.tight_layout()
plt.show()
print(f"Sent:     {MESSAGE.upper()!r}")
print(f"Received: {decoded!r}")
```

### How robust is the channel?

In the real world, a sticker is never placed at exactly 90°. We now treat the patches as a noisy communication channel and measure the **character error rate (CER)** of the received message:

* **Rotation noise:** each frame's angle is perturbed uniformly within ±J degrees. Training only used ±10°, and beyond ±45° a symbol can land in a neighboring key.
* **Scale noise:** the patch is shown smaller or larger than its training size (0.9–1.1).

We also try a **repetition code**: each character is sent on 3 or 5 different images and decoded by majority vote. This is the simplest form of error correction, and it costs more frames per character.

```python
def char_error_rate(sent, received):
    return sum(a != b for a, b in zip(sent.upper(), received)) / len(sent)


NUM_TRIALS = 20
REPEATS = [1, 3, 5]
jitters = [0, 10, 20, 30, 40, 50]
scales = [0.5, 0.6, 0.7, 0.8, 0.9, 1.0, 1.1, 1.2]

fig, ax = plt.subplots(1, 2, figsize=(12, 4))
for r in REPEATS:
    cer_rot = [
        np.mean(
            [
                char_error_rate(
                    MESSAGE, transmit(MESSAGE, repeats=r, angle_jitter=j)[0]
                )
                for _ in range(NUM_TRIALS)
            ]
        )
        for j in tqdm(jitters, leave=False)
    ]
    cer_scale = [
        np.mean(
            [
                char_error_rate(
                    MESSAGE, transmit(MESSAGE, repeats=r, scale_range=(s, s))[0]
                )
                for _ in range(NUM_TRIALS)
            ]
        )
        for s in tqdm(scales, leave=False)
    ]
    ax[0].plot(jitters, np.array(cer_rot) * 100, marker="o", label=f"{r}x repetition")
    ax[1].plot(scales, np.array(cer_scale) * 100, marker="o", label=f"{r}x repetition")
ax[0].axvspan(0, 10, color="C2", alpha=0.1, label="training range")
ax[0].set_xlabel("Rotation noise ±J (degrees)")
ax[1].axvspan(0.9, 1.1, color="C2", alpha=0.1, label="training range")
ax[1].set_xlabel("Patch scale")
for a in ax:
    a.set_ylabel("Character error rate (%)")
    a.set_ylim(-2, 102)
    a.legend()
ax[0].set_title("Robustness to rotation noise")
ax[1].set_title("Robustness to scale")
plt.tight_layout()
plt.show()
```

### Discussion

*(Fill in after running.)* Questions to answer from the results above:

* **Capacity:** How do the top-1 fooling rates of a 4-key patch compare to the single-class 64×64 patches from earlier (>99%)? Is one 64×64 patch big enough to hold four classes? Which keys were hardest, and is there an obvious reason (e.g. semantic similarity to the image content)?
* **Crosstalk:** When a rotation fails, does it trigger a *different key of the same patch*, or some unrelated class? High crosstalk means the four personalities share features; low crosstalk means they are cleanly separated by orientation.
* **Between the keys:** In the angle sweep, is the switch at 45° sharp or gradual? Is there a "dead zone" where the true class wins again? That tells us whether ResNet34 is rotation-sensitive for these patterns (CNNs are not rotation-equivariant by design).
* **Channel robustness:** Where does the CER start to rise compared to the ±10° / 0.9–1.1 training range? How much does the repetition code buy, and at what cost in frames?
* **Security:** The decoder is a public model and the rule is trivial, so this is *steganography* rather than encryption. Anyone who guesses "run it through a ResNet34" can read the message. What would change if the sender used a private fine-tuned model as the key?
